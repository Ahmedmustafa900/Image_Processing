#Importing OpenCV, NumPy, and Matplotlib Libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

# Displaying the Image

```python
image = cv2.imread(r'E:\python\story_lena_lenna_1.jpg')
```

```python
plt.figure(figsize=(15, 10))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

#Image Processing Filters and Effects Functions

```python
def apply_grayscale(image):
    return cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

def apply_blur(image):
    return cv2.GaussianBlur(image, (15, 15), 0)

def apply_sharpen(image):
    kernel = np.array([[-1, -1, -1],
                      [-1, 9, -1],
                      [-1, -1, -1]])
    return cv2.filter2D(image, -1, kernel)

def apply_edge_detection(image):
    return cv2.Canny(image, 100, 200)

def apply_invert_colors(image):
    return cv2.bitwise_not(image)


def apply_cartoonize(image):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    edges = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 9, 9)
    color = cv2.bilateralFilter(image, 9, 300, 300)
    return cv2.bitwise_and(color, color, mask=edges)

def apply_pixelate(image, scale=10):
    h, w = image.shape[:2]
    small = cv2.resize(image, (w // scale, h // scale))
    return cv2.resize(small, (w, h), interpolation=cv2.INTER_NEAREST)

def apply_thermal_vision(image):
    h, w, _ = image.shape
    thermal_image = np.zeros((h, w, 3), dtype=np.uint8)
    thermal_image[:, :, 0] = 255
    thermal_image[:, :, 1] = ((image[:, :, 1] + image[:, :, 2]) / 2).astype(np.uint8)
    thermal_image[:, :, 2] = image[:, :, 0]
    return thermal_image


def apply_vintage(image):
    vintage = cv2.applyColorMap(image, cv2.COLORMAP_PINK)
    return vintage

def apply_watercolor(image):
    kernel = np.array([[-1, -1, -1],
                      [-1, 8, -1],
                      [-1, -1, -1]], dtype=np.float32)
    watercolor = cv2.filter2D(image, -1, kernel)
    return watercolor

def apply_cinematic(image):
    color_corrected = cv2.applyColorMap(image, cv2.COLORMAP_WINTER)
    return cv2.addWeighted(image, 0.7, color_corrected, 0.3, 0)

def apply_pencil_sketch(image):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    gray_inv = cv2.bitwise_not(gray)
    comic = cv2.adaptiveThreshold(gray_inv, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 9, 9)
    return cv2.cvtColor(comic, cv2.COLOR_GRAY2BGR)

def apply_lomo(image):
    lomo = cv2.applyColorMap(image, cv2.COLORMAP_SUMMER)
    return lomo

def apply_rainbow(image):
    h, w, _ = image.shape
    rainbow = np.zeros((h, w, 3), dtype=np.uint8)
    rainbow[:, :, 0] = 255
    rainbow[:, :, 1] = np.sin(np.linspace(0, 2 * np.pi, w)) * 255
    rainbow[:, :, 2] = np.cos(np.linspace(0, 2 * np.pi, w)) * 255
    return cv2.addWeighted(image, 0.7, rainbow, 0.3, 0)
```

#Multi-Filter Image Processing Pipeline

```python
grayscale_image = apply_grayscale(image)

blurred_image = apply_blur(image)

sharpened_image = apply_sharpen(image)

edge_image = apply_edge_detection(image)

inverted_colors_image = apply_invert_colors(image)

cartoonized_image = apply_cartoonize(image)

pixelated_image = apply_pixelate(image)

thermal_vision_image = apply_thermal_vision(image)

vintage_image = apply_vintage(image)

watercolor_image = apply_watercolor(image)

cinematic_image = apply_cinematic(image)

pencil_sketch_image = apply_pencil_sketch(image)

lomo_image = apply_lomo(image)

rainbow_image = apply_rainbow(image)
```

#Image Processing Result Saving

```python
cv2.imwrite('grayscale_image.jpg', grayscale_image)
cv2.imwrite('blurred_image.jpg', blurred_image)
cv2.imwrite('sharpened_image.jpg', sharpened_image)
cv2.imwrite('edge_image.jpg', edge_image)
cv2.imwrite('inverted_colors_image.jpg', inverted_colors_image)
cv2.imwrite('cartoonized_image.jpg', cartoonized_image)
cv2.imwrite('pixelated_image.jpg', pixelated_image)
cv2.imwrite('thermal_vision_image.jpg', thermal_vision_image)
cv2.imwrite('vintage_image.jpg', vintage_image)
cv2.imwrite('watercolor_image.jpg', watercolor_image)
cv2.imwrite('cinematic_image.jpg', cinematic_image)
cv2.imwrite('pencil_sketch_image.jpg', pencil_sketch_image)
cv2.imwrite('lomo_image.jpg', lomo_image)
cv2.imwrite('rainbow_image.jpg', rainbow_image)
```

```python
images = [
    ('Original Image', image),
    ('Grayscale Image', grayscale_image),
    ('Cartoonize Image', cartoonized_image),
    ('Blurred Image', blurred_image),
    ('Sharpened Image', sharpened_image),
    ('Edge Detection', edge_image),
    ('Inverted Colors', inverted_colors_image),
    ('Pixelated Effect', pixelated_image),
    ('Thermal Vision Effect', thermal_vision_image),
    ('Vintage Effect', vintage_image),
    ('Watercolor Effect', watercolor_image),
    ('Cinematic Effect', cinematic_image),
    ('Pencil Sketch Image', pencil_sketch_image),
    ('Lomo Effect', lomo_image),
    ('Rainbow Effect', rainbow_image),
]

for title, img in images:
    plt.figure(figsize=(15, 10))
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.axis('off')
    plt.show()
```
